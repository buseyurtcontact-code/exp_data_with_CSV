# 4. Koşullar: Programın karar vermesi

Bir program her durumda aynı şeyi yapmak zorunda değildir.

Gerçek hayatta:
- Eğer yağmur yağıyorsa şemsiye al.
- Eğer not 50 veya üzeriyse geçti.
- Değilse kaldı.

Python'da bunun için `if`, `elif` ve `else` kullanılır.

```python
notu = 73

if notu >= 50:
    print("Sonuç: Başarılı")
else:
    print("Sonuç: Başarısız")
```

### Girinti neden önemli?

Python'da girinti, hangi kodun hangi koşula ait olduğunu gösterir.

Aşağıdaki örnekte birden fazla karar seviyesi var.

```python
notu = 88

if notu >= 85:
    harf = "AA"
elif notu >= 70:
    harf = "BB"
elif notu >= 55:
    harf = "CC"
else:
    harf = "FF"

print("Not:", notu)
print("Harf karşılığı:", harf)
```

### Deneyin
`notu = 88` değerini sırasıyla `72`, `58` ve `40` yapıp hücreyi tekrar çalıştırın.

Tek bir değer yerine bir grup değeri saklamak için liste kullanılabilir.

Örneğin bir gruptaki birkaç sınav notu:

```python
notlar = [72, 88, 65, 91, 54]

print("Tüm notlar:", notlar)
print("İlk not:", notlar[0])
print("Not sayısı:", len(notlar))
```

> **İki önemli kural:**
>- Python'da listedeki ilk elemanın sıra numarası (indeksi) `0`'dır. Bu nedenle `notlar[0]` ilk değeri verir.

# 6. Döngüler: Tekrar eden işleri otomatikleştirmek

Bir listedeki her değer için aynı işlemi yapmak istediğimizi düşünelim.

Tek tek yazmak yerine `for` döngüsü kullanabiliriz.

```python
notlar = [72, 88, 65, 91, 54]

for notu in notlar:
    print("İşlenen not:", notu)
```

Şimdi döngü ile koşulu birleştirelim.

Bu örnek, programlamanın temel mantığını çok iyi gösterir:
- Veriyi al,
- sırayla işle,
- her değer için karar ver,
- çıktı üret.

```python
notlar = [72, 88, 45, 91, 39]

for notu in notlar:
    if notu >= 50:
        print(notu, "→ geçti")
    else:
        print(notu, "→ kaldı")
```

### Ek kullanım senaryosu 1 — Katılımcılara kişisel karşılama mesajı

**Senaryo:** Bir atölyedeki her katılımcı için aynı yapıda bir mesaj göstermek istiyorsunuz. Mesajlar yalnızca ekrana yazılır.

```python
katilimcilar = ["Deniz", "Ece", "Mert"]

for katilimci in katilimcilar:
    print("Hoş geldin", katilimci, "! Atölye başlıyor.")
```

**Gözlem / beklenen sonuç:** Üç kişi için üç mesaj oluşur. Her turda katilimci değişkeni listedeki sıradaki ismi alır.

**Deneyin:** Listeye iki isim ekleyin. print satırını çoğaltmadan beş mesaj üretildiğini gözlemleyin.

### Ek kullanım senaryosu 2 — Düşük stokları saymak

**Senaryo:** Bir raftaki ürünlerin stok adetleri kontrol edilecek. Örnek kurala göre 5 adedin altındaki ürünlerin yenilenmesi gerekiyor.

```python
stok_adetleri = [12, 3, 0, 8, 4]
yenilenecek_urun_sayisi = 0

for stok in stok_adetleri:
    if stok < 5:
        print("Stok", stok, "→ yenileme gerekli")
        yenilenecek_urun_sayisi = yenilenecek_urun_sayisi + 1
    else:
        print("Stok", stok, "→ yeterli")

print("Yenilenecek ürün sayısı:", yenilenecek_urun_sayisi)
```

**Gözlem / beklenen sonuç:** 3, 0 ve 4 adetlik stoklar için sayaç artar; toplam 3 ürün yenilenmelidir. En alttaki print, girintisi olmadığı için döngü bittikten sonra bir kez çalışır.

**Deneyin:** Listeye 5 ekleyin. Sayaç artar mı? < ile <= arasındaki farkı bu değer üzerinden tartışın.

# 7. Fonksiyonlar: Bir işlemi paketlemek

Aynı işlemi tekrar tekrar kullanacaksak onu bir **fonksiyon** haline getirebiliriz.

Fonksiyonu küçük bir makine gibi düşünebilirsiniz:

**Girdi → Fonksiyon → Sonuç**

```python
def not_durumu(notu):
    if notu >= 50:
        return "Geçti"
    else:
        return "Kaldı"

print(not_durumu(72))
print(not_durumu(43))
```

Burada:
- `def` ile fonksiyon tanımlandı.
- `notu` fonksiyona verilen girdidir.
- `return` fonksiyonun ürettiği sonucu dışarı verir.

Fonksiyonu bir kez tanımlayıp birçok farklı veri için kullanabiliriz.

```python
ornek_notlar = [90, 48, 67, 35, 81]

for notu in ornek_notlar:
    sonuc = not_durumu(notu)
    print(notu, "→", sonuc)
```

### Ek kullanım senaryosu 1 — Dakikayı saniyeye dönüştüren araç

**Senaryo:** Farklı etkinlik sürelerini saniye cinsinden hesaplamak için aynı dönüşümü tekrar kullanacaksınız.

```python
def saniyeye_cevir(dakika):
    return dakika * 60

print("5 dakika:", saniyeye_cevir(5), "saniye")
print("12 dakika:", saniyeye_cevir(12), "saniye")
toplam_saniye = saniyeye_cevir(3) + saniyeye_cevir(2)
print("İki etkinliğin toplamı:", toplam_saniye, "saniye")
```

**Gözlem / beklenen sonuç:** Çıktılar sırasıyla 300, 720 ve 300 saniyedir. return ile dönen sayılar başka bir hesapta da kullanılabilir.

**Deneyin:** Fonksiyonu 1.5 girdisiyle çağırın. 90 saniye sonucunu elde edin.

### Ek kullanım senaryosu 2 — Dikdörtgen pano alanını hesaplamak

**Senaryo:** Okuldaki farklı panoların alanları hesaplanacak. Bu fonksiyon virgülle ayrılmış iki girdi alır: genişlik ve yükseklik. Ölçüler santimetredir.

```python
def pano_alani(genislik, yukseklik):
    alan = genislik * yukseklik
    return alan

kucuk_pano = pano_alani(40, 30)
buyuk_pano = pano_alani(80, 50)
print("Küçük pano:", kucuk_pano, "cm²")
print("Büyük pano:", buyuk_pano, "cm²")
```

**Gözlem / beklenen sonuç:** Alanlar 1200 ve 4000 cm² bulunur. def işlemi tanımlar; hesap, fonksiyon çağrıldığında verilen ölçülerle yapılır.

**Deneyin:** 60 × 40 cm ölçülü üçüncü bir pano için çağrı ekleyin. İşlemi fonksiyonun dışında yeniden yazmanız gerekiyor mu?

# 8. Kütüphaneler: Hazır araç kutuları

Python'un önemli güçlerinden biri kütüphanelerdir.

Kütüphane, başkaları tarafından hazırlanmış fonksiyon ve araçları kullanmamızı sağlar.

`import` ifadesini **"bu araç kutusunu çalışma alanına getir"** şeklinde düşünebilirsiniz.

```python
import math
import statistics

sayilar = [10, 20, 30, 40, 50]

print("Karekök:", math.sqrt(81))
print("Ortalama:", statistics.mean(sayilar))
print("Ortanca:", statistics.median(sayilar))
```

## Neden kütüphane kullanıyoruz?

Çünkü her şeyi sıfırdan yazmak istemeyiz.

Örneğin:
- `math` → matematik işlemleri,
- `statistics` → temel istatistik,
- `pandas` → tablo ve veri analizi,
- `matplotlib` → grafik,
- `numpy` → sayısal işlemler.

### Ek kullanım senaryosu 1 — Gezi için araç sayısını hesaplamak

# 9. Tablo mantığı: `pandas`

Gerçek hayattaki veriler çoğu zaman Excel benzeri satır ve sütunlardan oluşur.

Python'da bu tür tablolarla çalışmak için en yaygın araçlardan biri `pandas` kütüphanesidir.

```python
import pandas as pd

veri = {
    "Öğrenci": ["Öğrenci 1", "Öğrenci 2", "Öğrenci 3", "Öğrenci 4", "Öğrenci 5"],
    "Not": [78, 45, 92, 66, 84]
}

tablo = pd.DataFrame(veri)
tablo
```

Colab, bir `DataFrame` nesnesini son satıra yazdığımızda tablo biçiminde gösterir.

Şimdi tabloya yeni bir sütun ekleyelim.

```python
def basari_durumu(notu):
    if notu >= 50:
        return "Geçti"
    else:
        return "Kaldı"

# apply() fonksiyonu, hazırladığımız kuralı tablodaki tüm satırlara sırayla otomatik uygular:
tablo["Durum"] = tablo["Not"].apply(basari_durumu)
tablo
```

Bu hücrede daha önce öğrendiğimiz kavramlar birleşti:

- `tablo` → değişken
- `Not` → veri
- `basari_durumu()` → fonksiyon
- `if / else` → koşul
- `apply()` → işlemi her satıra uygulama
- sonuç → yeni sütun

# 11. Küçük bir uçtan uca örnek

Şimdi temel kavramları tek bir akışta birleştirelim.

### Senaryo
Bir sınıfın puanlarını:
1. değişkende tutacağız,
2. fonksiyonla sınıflandıracağız,
3. tablo oluşturacağız,
4. özet bilgi üreteceğiz,
5. grafikle göstereceğiz.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Örnek veri grupları
ogrenciler = ["Öğrenci A", "Öğrenci B", "Öğrenci C", "Öğrenci D", "Öğrenci E", "Öğrenci F"]
puanlar = [42, 76, 91, 58, 67, 35]

def seviye_belirle(puan):
    if puan >= 85:
        return "Yüksek"
    elif puan >= 50:
        return "Orta"
    else:
        return "Destek Gerekli"

sonuclar = []

# zip() fonksiyonu iki listeyi fermuar gibi birleştirip (isim ve not) ikilisini birlikte almamızı sağlar:
for ad, puan in zip(ogrenciler, puanlar):
    sonuclar.append({
        "Öğrenci": ad,
        "Puan": puan,
        "Seviye": seviye_belirle(puan)
    })

sinif = pd.DataFrame(sonuclar)
sinif
```

```python
sinif_ortalamasi = sinif["Puan"].mean()
gecen_sayisi = (sinif["Puan"] >= 50).sum()

print(f"Sınıf ortalaması: {sinif_ortalamasi:.1f}")
print(f"50 ve üzeri puan alan öğrenci sayısı: {gecen_sayisi}")
```

```python
plt.figure(figsize=(8, 4))
plt.bar(sinif["Öğrenci"], sinif["Puan"])
plt.axhline(50, linestyle="--", color="red", label="50 puan sınırı")
plt.title("Sınıf Puan Dağılımı")
plt.xlabel("Öğrenci")
plt.ylabel("Puan")
plt.ylim(0, 100)
plt.legend()
plt.show()
```

## Bu örnekte akış neydi?

**Veri → Fonksiyon → Döngü → Koşul → Tablo → Hesap → Görselleştirme**

Gerçek veri bilimi, yapay zekâ ve otomasyon çalışmalarında kodlar çok daha uzun olabilir; fakat temel akış çoğu zaman buna benzer.

### Ek kullanım senaryosu 1 — Kitap okuma hedefini izlemek

**Senaryo:** Bir kulüpte üç öğrencinin okuduğu sayfaları kaydedip 100 sayfalık örnek hedefe ulaşıp ulaşmadığını bulacaksınız. Üç kod hücresini sırayla çalıştırın.

```python
import pandas as pd
import matplotlib.pyplot as plt

okuyanlar = ["Deniz", "Ece", "Mert"]
okunan_sayfalar = [80, 120, 100]
sayfa_hedefi = 100

def okuma_durumu(sayfa, hedef):
    if sayfa >= hedef:
        return "Hedefe ulaştı"
    else:
        return "Devam ediyor"

okuma_sonuclari = []
for kisi, sayfa in zip(okuyanlar, okunan_sayfalar):
    okuma_sonuclari.append({
        "Öğrenci": kisi,
        "Sayfa": sayfa,
        "Durum": okuma_durumu(sayfa, sayfa_hedefi)
    })
okuma_tablosu = pd.DataFrame(okuma_sonuclari)
okuma_tablosu
```

**Özet üretelim:** sum() sütundaki sayıları toplar; koşulun ardından kullanılan sum(), koşulu sağlayan satırları sayar.

```python
print("Toplam okunan sayfa:", okuma_tablosu["Sayfa"].sum())
print("Hedefe ulaşan öğrenci:", (okuma_tablosu["Sayfa"] >= sayfa_hedefi).sum())
```

```python
plt.figure(figsize=(8, 4))
plt.bar(okuma_tablosu["Öğrenci"], okuma_tablosu["Sayfa"])
plt.axhline(sayfa_hedefi, color="red", linestyle="--", label="Sayfa hedefi")
plt.title("Okuma Kulübünde Hedef Takibi")
plt.xlabel("Öğrenci")
plt.ylabel("Okunan sayfa sayısı")
plt.ylim(bottom=0)
plt.legend()
plt.tight_layout()
plt.show()
```

**Gözlem / beklenen sonuç:** Toplam 300 sayfa okunmuştur; Ece ve Mert hedefe ulaşır. Aynı sayfa_hedefi değeri sınıflandırma, özet ve grafikte kullanılır.

**Deneyin:** İlk hücrede hedefi 90 yapın ve üç hücreyi yeniden sırayla çalıştırın. Sonra Deniz’in sayfa sayısını 95 yapıp tekrar deneyin.

### Ek kullanım senaryosu 1 — Metin olarak gelen adedi sayıya çevirmek

**Senaryo:** Bir formdan gelen bilet adedi "üç" metni olarak kaydedilmiş. int() sayısal metni tam sayıya çevirir; bu yazıyla belirtilmiş değeri çeviremez. Hatayı yakalayıp ardından geçerli örneği göstereceksiniz.

```python
adet_metni = "üç"

try:
    bilet_adedi_sayi = int(adet_metni)
    print("Bilet adedi:", bilet_adedi_sayi)
except ValueError as hata:
    print("Hata türü:", type(hata).__name__)
    print("Mesaj:", hata)
    print("Adedi rakamlarla yazın; örneğin 3.")

duzeltilmis_adet_metni = "3"
print("Düzeltilmiş giriş:", int(duzeltilmis_adet_metni))
```

**Gözlem / beklenen sonuç:** İlk giriş ValueError üretir ve except bloğu bunu açıklar. "3" metni ise 3 sayısına dönüşür. Koruma sayesinde sonraki satırlar çalışmaya devam eder.

**Deneyin:** İlk girdiyi "5" yapın. except bloğu çalışır mı? Ardından "3.5" deneyip int() için bunun da geçerli tam sayı metni olmadığını görün.

### Ek kullanım senaryosu 2 — Listede olmayan sıraya erişmek

**Senaryo:** Üç oturumluk bir programdan dördüncü oturumu okumaya çalışıyorsunuz. Önce hatayı görün, sonra geçerli indeksle erişin.

```python
oturumlar = ["Tanışma", "Uygulama", "Değerlendirme"]

try:
    print(oturumlar[3])
except IndexError as hata:
    print("Hata türü:", type(hata).__name__)
    print("Mesaj:", hata)
    print("Üç elemanlı bu listenin indeksleri 0, 1 ve 2'dir.")

print("Üçüncü oturum:", oturumlar[2])
```

**Gözlem / beklenen sonuç:** İndeks 3 dördüncü elemanı istediği için IndexError oluşur. İndeks 2 kullanıldığında Değerlendirme yazdırılır.

### Ek kullanım senaryosu 1 — Etkinlik 7: Atölye malzemesi planlamak

**Senaryo:** Bir atölyede her öğrenciye iki kalem verilecek. Eldeki 25 kalemin 12 öğrenciye yetip yetmediğini kontrol edin. Önce sonucu tahmin edin, sonra kodu çalıştırın.

```python
atolye_ogrenci_sayisi = 12
kisi_basi_kalem = 2
mevcut_kalem = 25
gereken_kalem = atolye_ogrenci_sayisi * kisi_basi_kalem

print("Gereken kalem:", gereken_kalem)
if mevcut_kalem >= gereken_kalem:
    print("Kalemler yeterli.")
    print("Artan kalem:", mevcut_kalem - gereken_kalem)
else:
    print("Ek kalem gerekli:", gereken_kalem - mevcut_kalem)
```

**Gözlem / beklenen sonuç:** 24 kalem gerekir; 25 kalem yeterlidir ve 1 kalem artar. Bu etkinlik değişken, işlem ve koşulu birleştirir.

**Deneyin:** Öğrenci sayısını 15 yapın; 5 kalem eksik olmalı. Sonra mevcut kalemi 30 yapın; eşitlik durumunda hangi mesajın çıkacağını tahmin edin.

### Ek kullanım senaryosu 2 — Etkinlik 8: Okuma günlerini saymak

**Senaryo:** Bir öğrencinin dört günlük okuma kaydı var. Günlük hedefe ulaşılan günleri döngüyle sayın; sonra hedefi değiştirerek sonucu karşılaştırın.

```python
gunluk_sayfalar = [10, 25, 20, 15]
gunluk_hedef = 20
hedefe_ulasilan_gun = 0

for sayfa in gunluk_sayfalar:
    if sayfa >= gunluk_hedef:
        hedefe_ulasilan_gun = hedefe_ulasilan_gun + 1

print("Kayıtlı gün:", len(gunluk_sayfalar))
print("Hedefe ulaşılan gün:", hedefe_ulasilan_gun)
```

**Gözlem / beklenen sonuç:** Dört günün ikisinde (25 ve 20 sayfa) hedefe ulaşılır. Sayaç her başarılı günde bir artar.

**Deneyin:** Hedefi 15 yapın: sonuç 3 olmalı. Sonra listeye 30 ekleyin: kayıtlı gün 5, hedefe ulaşılan gün 4 olmalı.

### Ek kullanım senaryosu 1 — Bir rezervasyon akışını kavramlarla eşleştirmek

**Senaryo:** Bir salonda 30 yerin 26’sı dolu. Gelen 3 kişilik talep için boş yer kontrol edilecek. Kodu okurken her satırın girdi, işlem, karar veya çıktı rolünü belirleyin.

```python
# Girdi: kapasite, dolu yer ve talep
salon_kapasitesi = 30
dolu_yer = 26
talep_edilen_yer = 3

# İşlem: kullanılabilir yeri bul
bos_yer = salon_kapasitesi - dolu_yer

# Karar ve çıktı
if talep_edilen_yer <= bos_yer:
    print("Talep karşılanabilir.")
    print("Talep karşılanırsa kalan yer:", bos_yer - talep_edilen_yer)
else:
    print("Yeterli boş yer yok.")
```

**Gözlem / beklenen sonuç:** 4 boş yer vardır; 3 kişilik talep karşılanabilir ve 1 yer kalır. Akış: veri → değişken → çıkarma → karşılaştırma → çıktı. Kod yalnızca uygunluğu kontrol eder, gerçek rezervasyon oluşturmaz.

**Deneyin:** Talebi 5 yapın. Hangi girdinin değiştiğini, hangi kararın değiştiğini ve çıktının nasıl değiştiğini ayrı ayrı söyleyin.

### Ek kullanım senaryosu 2 — Tekrarlanan dönüşümü fonksiyonla okumak

```python
def video_saniyesi(dakika):
    return dakika * 60

video_dakikalari = [2, 5, 3]
for sure in video_dakikalari:
    print(sure, "dakika →", video_saniyesi(sure), "saniye")
```

# 16. Kapanış: Kodun tamamını anlamak zorunda değilsiniz

Bir Python çalışmasını incelerken her komutu ezberlemek yerine önce büyük resmi görün:

> **Bu kod hangi veriyi alıyor, üzerinde ne yapıyor ve sonunda ne üretiyor?**

Bu soruyu cevaplayabiliyorsanız, programlama akışını anlamaya başlamışsınız demektir.

### Ek kullanım senaryosu 1 — Tanımadığınız bir kodda büyük resmi görmek

**Senaryo:** Bir arkadaşınız paylaşılan aşağıdaki kodun ne yaptığını soruyor. Önce yalnızca değişken isimlerine ve çıktıya bakarak tek cümlelik bir açıklama üretin.

```python
import statistics

gunluk_adimlar = [4000, 6000, 5000]
ortalama_adim = statistics.mean(gunluk_adimlar)
print("Ortalama adım sayısı:", ortalama_adim)
```

**Gözlem / beklenen sonuç:** Kod üç günlük adım sayısını alır, ortalamasını hesaplar ve 5000 sonucunu gösterir. Büyük resim: adım kayıtları → hazır ortalama aracı → tek bir özet sayı.

**Deneyin:** Listeye 9000 eklemeden önce ortalamanın hangi yönde değişeceğini söyleyin. Sonra çalıştırıp 6000 sonucuyla açıklamanızı kontrol edin.

### Ek kullanım senaryosu 2 — Yeni bir senaryoya öğrendiklerini taşımak

**Senaryo:** Dersin sonunda artık bir ölçüm listesinin hangi amaçla işlendiğini okuyabilirsiniz. Bu örnekte oda sıcaklığı için belirlenen 25 °C örnek sınırının aşıldığı ölçümler sayılıyor; bu sınır yalnızca alıştırma içindir.

```python
oda_olcumleri = [22, 27, 24, 29]
ornek_sinir = 25
siniri_asan_olcum = 0

for sicaklik_olcumu in oda_olcumleri:
    if sicaklik_olcumu > ornek_sinir:
        siniri_asan_olcum = siniri_asan_olcum + 1

print("Sınırı aşan ölçüm sayısı:", siniri_asan_olcum)
```

**Gözlem / beklenen sonuç:** 27 ve 29 sınırı aştığı için sonuç 2 olur. Büyük resim: ölçümler → tek tek karşılaştırma → sayma → özet çıktı.

**Deneyin:** Sınırı 28 yapın ve sonucu tahmin edin. Ardından şu cümleyi kendi sözlerinizle tamamlayın: “Bu kod … verisini alıyor, … işlemini yapıyor ve … üretiyor.”
